```python
test_settings = {'language': 'ehglish',
                'template': 'white'}
def add_setting(set_dict, add_tuple):
    
    add_list = [i.lower() for i in add_tuple]
    add_tuple = tuple(add_list)
    
    key = add_tuple[0]
    value = add_tuple[1]

    set_keys_list = list(set_dict.keys())
    if key in set_keys_list:
        return f"Setting '{key}' already exists! Cannot add a new setting with this name."
    else:
        set_dict.update({f'{key}': f'{value}'})
        return f"Setting '{key}' added with value '{value}' successfully!"

def update_setting(set_dict, update_tuple):
    update_list = [i.lower() for i in update_tuple]
    update_tuple = tuple(update_list)
    
    key = update_tuple[0]
    value = update_tuple[1]
    
    set_keys_list = list(set_dict.keys())
    if key in set_keys_list:
        set_dict.update({f'{key}': f'{value}'})
        return f"Setting '{key}' updated to '{value}' successfully!"
    else:
        return f"Setting '{key}' does not exist! Cannot update a non-existing setting." 

def delete_setting(set_dict, key):
    key = key.lower()

    set_keys_list = list(set_dict.keys())
    if key in set_keys_list:
        set_dict.pop(key)
        return f"Setting '{key}' deleted successfully!"
    else:
        return "Setting not found!"

def view_settings(set_dict):
    if not set_dict:
        return 'No settings available.'
    else:
        view = 'Current User Settings:\n'
        for key, value in set_dict.items():
            view += f'{key.title()}: {value}\n'
        return view
```
