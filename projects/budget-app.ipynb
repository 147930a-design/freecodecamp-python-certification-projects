```python
class Category:
    def __init__(self, name):
        self.name = name
        self.ledger = []

    def check_funds(self, money):
        funds = 0
        for i in self.ledger:
            funds += i['amount'] 
        
        if money > funds:
            return False
        else:
            return True

    def deposit(self, money, description=''):
        self.ledger.append(
            {'amount': money,
            'description': description})
    
    def withdraw(self, money, description=''):
        if not self.check_funds(money):
            return False
        else:
            self.ledger.append(
                {'amount': -money,
                 'description': description})
            return True

    def get_balance(self):
        balance = 0
        for i in self.ledger:
            balance += i['amount'] 
        return balance

    def transfer(self, money, category):
        if not self.check_funds(money):
            return False
        else:
            self.ledger.append(
                {'amount': -money,
                'description': f'Transfer to {category.name}'})

            category.ledger.append(
                {'amount': money,
                'description': f'Transfer from {self.name}'})
            return True

    def __str__(self):
        strs = self.name.center(30, '*') + '\n'
        for dicy_idx in self.ledger:
            description = dicy_idx['description'][:23].ljust(23)
            amount = f'{dicy_idx["amount"]:.2f}'[:7].rjust(7)
            strs += description + amount + '\n'
        strs += f'Total: {self.get_balance()}'
        return strs


def create_spend_chart(categories):
    chart = 'Percentage spent by category\n'

    spent_amounts = []
    for category in categories:
        spent = sum(abs(item['amount'])
            for item in category.ledger if item['amount'] < 0)
        spent_amounts.append(spent)

    total_spent = sum(spent_amounts)
    Percentage_spent = [int(s / total_spent * 100 // 10) * 10 for s in spent_amounts]
    
    for i in range(100, -1, -10):
        chart = chart + str(i).rjust(3) + '| '
        for idx in Percentage_spent:
            if idx >= i:
                chart += 'o  '
            else:
                chart += '   '
        chart += '\n'
     
    line = '    ' + '-'*(3*len(Percentage_spent)+1) + '\n'

    name_len = max(len(category.name) for category in categories) 
    name_lines = []
    for i in range(name_len):
        names = '     '
        for category in categories:
            if len(category.name) <= i:
                names += '   '
            else:
                names += category.name[i] + '  '
        name_lines.append(names)
    
    return chart + line + '\n'.join(name_lines)
```
